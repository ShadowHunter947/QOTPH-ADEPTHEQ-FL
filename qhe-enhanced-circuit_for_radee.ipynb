```python
# ======================================================================================
# LazyQ-HRU v4 (tau trigger fixed: stale-gradient stepping)  |  MNIST, one cell
# --------------------------------------------------------------------------------------
# Implements, faithfully:
#   QuCNet  Eq.13    : QCM measurement -> full 2^n basis-state probabilities per circuit
#   QuCNet  Eq.15-16 : HCWS weight sharing (stride: circuit k uses set k mod R)
#   QuCNet  Table 4  : with-TQC vs without-TQC ablation
#   QuCNet  Table 5  : weight-sharing ablation (naive/full training, else confounded)
#   PHQC-Net Eq.20   : data re-uploading (fresh feature block per layer, RY enc, ring CNOT)
#   PHQC-Net metric  : branch feature-energy contribution ratio
#   Ours             : lazy refresh with STALE-GRADIENT stepping -> tau dial is live
# --------------------------------------------------------------------------------------
# v4 changes vs v3 (grounded in v3 logs):
#   * v3 bug: q_weights only updated inside refresh steps -> drift==0 between refreshes
#     -> `drift > tau` never fired; Q-bwd == forced schedule only (10/376, 5/188);
#     TAU sweep flat. FIX: on lazy steps, apply the cached STALE quantum gradients
#     (oq.step() with grads from the last fresh backward). Weights drift every step,
#     tau triggers real refreshes, and the setup matches stale-SGD convergence analysis.
#   * v3 bug: `torch.no_grad()` on lazy steps also froze the adapters. FIX: adapters are
#     classical params and now receive gradient every step (quantum output detached only
#     at the circuit boundary on lazy steps).
#   * Ablation 1 (sharing) now runs naive (lazy=False): comparing sharing schemes with
#     ~5 quantum updates measures nothing.
# Prototype for Radee to validate before results are circulated.
# ======================================================================================

import os, math, time, random, warnings
warnings.filterwarnings("ignore")
import numpy as np
import torch, torch.nn as nn, torch.nn.functional as F
from torch.utils.data import DataLoader, Subset
from torchvision import datasets, transforms
from sklearn.metrics import accuracy_score, f1_score
import pennylane as qml

# -- 1. REPRO ---------------------------------------------------------------------------
SEED = 7
def set_seed(s=SEED):
    random.seed(s); np.random.seed(s); torch.manual_seed(s); torch.cuda.manual_seed_all(s)
set_seed()
DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"[env] device={DEVICE} | torch={torch.__version__} | pennylane={qml.__version__}")

# -- 2. CONFIG ---------------------------------------------------------------------------
N_TRAIN_SUB, N_TEST_SUB = 6000, 2000
BATCH_SIZE, LR          = 128, 1e-3
LATENT_DIM, N_CLASSES   = 16, 10

N_QUBITS, K_CIRCUITS, R_UNIQUE, RU_LAYERS = 4, 8, 4, 3
MEASURE = "probs"        # "probs" = QuCNet QCM (2^n dims/circuit) | "expval" = <Z> (n dims)

TAU_MAIN      = 0.15     # staleness threshold on ||theta_q - anchor||
REFRESH_EVERY = 100      # hard cap on staleness (safety net, should rarely bind now)
MAX_REFRESH   = 400

MAIN_EPOCHS     = 8
ABLATION_EPOCHS = 4
TAU_SWEEP       = [0.05, 0.15, 0.40]
SHARE_MODES     = ["hcws", "all_shared", "all_independent", "alternate"]

QDIM = (2**N_QUBITS) if MEASURE == "probs" else N_QUBITS
print(f"[cfg] qubits={N_QUBITS} K={K_CIRCUITS} R={R_UNIQUE} RU={RU_LAYERS} "
      f"measure={MEASURE} (qdim/circuit={QDIM}) tau={TAU_MAIN}")

# -- 3. CIRCUIT (PHQC-Net Eq.20 re-uploading; QuCNet Eq.13 QCM measurement) --------------
# NOTE: adjoint diff supports expectation values only; qml.probs needs backprop.
_dev_adj = qml.device("lightning.qubit", wires=N_QUBITS)   # fast adjoint for <Z>
_dev_bp  = qml.device("default.qubit",  wires=N_QUBITS)    # backprop for probs (QCM)

@qml.qnode(_dev_adj, interface="torch", diff_method="adjoint")
def circuit_expval(inputs, weights):
    for l in range(RU_LAYERS):
        qml.AngleEmbedding(inputs[:, l*N_QUBITS:(l+1)*N_QUBITS],
                           wires=range(N_QUBITS), rotation="Y")     # D_l fresh block
        for q in range(N_QUBITS): qml.RY(weights[l, q], wires=q)    # R(theta_l)
        for q in range(N_QUBITS-1): qml.CNOT(wires=[q, q+1])        # E_n ring
        qml.CNOT(wires=[N_QUBITS-1, 0])
    return [qml.expval(qml.PauliZ(q)) for q in range(N_QUBITS)]

@qml.qnode(_dev_bp, interface="torch", diff_method="backprop")
def circuit_probs(inputs, weights):
    for l in range(RU_LAYERS):
        qml.AngleEmbedding(inputs[:, l*N_QUBITS:(l+1)*N_QUBITS],
                           wires=range(N_QUBITS), rotation="Y")
        for q in range(N_QUBITS): qml.RY(weights[l, q], wires=q)
        for q in range(N_QUBITS-1): qml.CNOT(wires=[q, q+1])
        qml.CNOT(wires=[N_QUBITS-1, 0])
    return qml.probs(wires=range(N_QUBITS))                          # QuCNet Eq.13 p(x)

def run_circuit(inputs, weights):
    # default.qubit / lightning.qubit hold the state on CPU: keep circuit inputs and
    # trainable weights on CPU (differentiable .cpu()); QBank ships outputs back to GPU.
    inputs  = inputs.to("cpu")
    weights = weights.to("cpu")
    if MEASURE == "probs":
        return circuit_probs(inputs, weights).float()                # [B, 2^n]
    out = circuit_expval(inputs, weights)
    return torch.stack(out, dim=1).float()                           # [B, n]

# -- 4. MODEL ------------------------------------------------------------------------------
class TinyCNN(nn.Module):
    def __init__(self, d=LATENT_DIM):
        super().__init__()
        self.body = nn.Sequential(
            nn.Conv2d(1, 8, 3, padding=1),  nn.ReLU(), nn.MaxPool2d(2),
            nn.Conv2d(8, 16, 3, padding=1), nn.ReLU(), nn.MaxPool2d(2),
            nn.AdaptiveAvgPool2d(1))
        self.proj = nn.Linear(16, d)
    def forward(self, x): return torch.tanh(self.proj(self.body(x).flatten(1)))

class Head(nn.Module):
    def __init__(self, ind):
        super().__init__()
        self.net = nn.Sequential(nn.Linear(ind, 64), nn.LayerNorm(64), nn.GELU(),
                                 nn.Dropout(0.1), nn.Linear(64, N_CLASSES))
    def forward(self, x): return self.net(x)

def share_id(k, mode):
    """QuCNet Table-5 weight-sharing schemes."""
    if mode == "all_shared":      return 0                       # q1->q1 x K   (R=1)
    if mode == "all_independent": return k                       # q1->...->qK  (R=K)
    if mode == "alternate":       return k // max(K_CIRCUITS // R_UNIQUE, 1)  # blocked
    return k % R_UNIQUE                                          # hcws (stride) Eq.16

class QBank(nn.Module):
    def __init__(self, mode="hcws", d=LATENT_DIM):
        super().__init__()
        self.mode = mode
        n_sets = {"all_shared": 1, "all_independent": K_CIRCUITS}.get(mode, R_UNIQUE)
        fpc = RU_LAYERS * N_QUBITS
        self.adapters  = nn.ModuleList([nn.Linear(d, fpc) for _ in range(K_CIRCUITS)])
        self.q_weights = nn.ParameterList([nn.Parameter(0.1*torch.randn(RU_LAYERS, N_QUBITS))
                                           for _ in range(n_sets)])
    def forward(self, latent, detach_circuit=False):
        """detach_circuit=True (lazy step): adapters still feed the circuit but the
        circuit runs under no_grad on detached inputs -> no quantum backward; the
        classical graph (CNN -> adapters) is untouched by the detach (adapters get
        gradient via the fused head only through the latent path; circuit output is
        treated as a constant feature)."""
        outs = []
        for k in range(K_CIRCUITS):
            feats = torch.tanh(self.adapters[k](latent)) * math.pi
            if detach_circuit:
                with torch.no_grad():
                    zk = run_circuit(feats.detach(), self.q_weights[share_id(k, self.mode)])
            else:
                zk = run_circuit(feats, self.q_weights[share_id(k, self.mode)])
            outs.append(zk.to(latent.device))
        return torch.cat(outs, dim=1)                            # [B, K*QDIM]
    def flat_weights(self):
        return torch.cat([w.detach().reshape(-1) for w in self.q_weights])

class Hybrid(nn.Module):
    def __init__(self, mode="hcws"):
        super().__init__()
        self.cnn = TinyCNN(); self.qbank = QBank(mode)
        self.head = Head(LATENT_DIM + K_CIRCUITS * QDIM)
    def latent(self, x): return self.cnn(x)
    def quantum(self, l, detach_circuit=False): return self.qbank(l, detach_circuit)
    def classify(self, l, q): return self.head(torch.cat([l, q], dim=1))

class Classical(nn.Module):
    def __init__(self):
        super().__init__(); self.cnn = TinyCNN(); self.head = Head(LATENT_DIM)

# -- 5. DATA -------------------------------------------------------------------------------
print("\n[data] loading MNIST ...")
tf = transforms.Compose([transforms.ToTensor(), transforms.Normalize((0.1307,), (0.3081,))])
_tr = datasets.MNIST("./data", train=True,  download=True, transform=tf)
_te = datasets.MNIST("./data", train=False, download=True, transform=tf)
g = torch.Generator().manual_seed(SEED)
train_loader = DataLoader(Subset(_tr, torch.randperm(len(_tr), generator=g)[:N_TRAIN_SUB].tolist()),
                          batch_size=BATCH_SIZE, shuffle=True)
test_loader  = DataLoader(Subset(_te, torch.randperm(len(_te), generator=g)[:N_TEST_SUB].tolist()),
                          batch_size=BATCH_SIZE, shuffle=False)
print(f"[data] train={N_TRAIN_SUB} test={N_TEST_SUB} batch={BATCH_SIZE}")

# -- 6. SANITY CHECKS ------------------------------------------------------------------------
print("\n[sanity-1] batched circuit + sharing wiring ...")
_b = QBank("hcws").to(DEVICE)
_l = torch.randn(BATCH_SIZE, LATENT_DIM, device=DEVICE, requires_grad=True)
_t = time.time(); _q = _b(_l); _dt = time.time() - _t
assert _q.shape == (BATCH_SIZE, K_CIRCUITS * QDIM), _q.shape
_q.sum().backward()
print(f"          out={tuple(_q.shape)} in {_dt*1000:.0f} ms | "
      f"hcws map={[share_id(k,'hcws') for k in range(K_CIRCUITS)]} | "
      f"alt map={[share_id(k,'alternate') for k in range(K_CIRCUITS)]}")
del _b, _l, _q

print("\n[sanity-2] Lipschitz premise (tau-dial reliability) ...")
_p = Hybrid("hcws").to(DEVICE).eval()
_xb = next(iter(test_loader))[0].to(DEVICE)
with torch.no_grad():
    _l0 = _p.latent(_xb); _q0 = _p.quantum(_l0)
    _w0 = [w.clone() for w in _p.qbank.q_weights]
    print(f"          {'eps':>6} | {'||dW||':>8} | {'||dFeat||':>9} | {'ratio~Lq':>8}")
    for eps in [0.01, 0.1, 0.5]:
        for w, wo in zip(_p.qbank.q_weights, _w0): w.copy_(wo + eps*torch.randn_like(wo))
        dW = math.sqrt(sum(((w-wo)**2).sum().item() for w,wo in zip(_p.qbank.q_weights,_w0)))
        dF = (_p.quantum(_l0) - _q0).norm().item()
        print(f"          {eps:>6.2f} | {dW:>8.3f} | {dF:>9.3f} | {dF/max(dW,1e-9):>8.3f}")
    for w, wo in zip(_p.qbank.q_weights, _w0): w.copy_(wo)
del _p

# -- 7. TRAIN / EVAL HELPERS ------------------------------------------------------------------
def evaluate(model, hybrid):
    model.eval(); ys, ps = [], []
    with torch.no_grad():
        for x, y in test_loader:
            x = x.to(DEVICE)
            logits = (model.classify(model.latent(x), model.quantum(model.latent(x)))
                      if hybrid else model.head(model.cnn(x)))
            ps.append(logits.argmax(1).cpu().numpy()); ys.append(y.numpy())
    ys, ps = np.concatenate(ys), np.concatenate(ps)
    return accuracy_score(ys, ps), f1_score(ys, ps, average="macro")

def contribution_ratio(model):
    """PHQC-Net-style feature-energy contribution: quantum share of fused feature energy."""
    model.eval(); e_lat = e_q = 0.0
    with torch.no_grad():
        for x, _ in test_loader:
            lat = model.latent(x.to(DEVICE)); qf = model.quantum(lat)
            e_lat += (lat**2).sum().item(); e_q += (qf**2).sum().item()
    return e_q / (e_q + e_lat)

def train_classical(epochs=MAIN_EPOCHS, verbose=True):
    set_seed(); m = Classical().to(DEVICE)
    opt = torch.optim.AdamW(m.parameters(), lr=LR); t0 = time.time()
    for ep in range(epochs):
        m.train()
        for x, y in train_loader:
            x, y = x.to(DEVICE), y.to(DEVICE)
            loss = F.cross_entropy(m.head(m.cnn(x)), y)
            opt.zero_grad(); loss.backward(); opt.step()
    acc, f1 = evaluate(m, hybrid=False)
    if verbose: print(f"    classical: acc={acc:.4f} f1={f1:.4f} ({time.time()-t0:.0f}s)")
    return dict(acc=acc, f1=f1, t=time.time()-t0, qb=0, model=m)

def train_hybrid(mode="hcws", tau=TAU_MAIN, epochs=MAIN_EPOCHS, lazy=True, verbose=True):
    """lazy=False -> naive hybrid: quantum backward EVERY step (QuCNet-style baseline).
    lazy=True  -> stale-gradient lazy training:
        refresh step : full quantum backward; cache fresh q-grads; step both optimizers.
        lazy step    : classical backward only; apply CACHED STALE q-grads via oq.step()
                       so theta_q keeps drifting -> ||theta_q - anchor|| > tau actually
                       triggers refreshes (v3's frozen-weights bug made tau inert)."""
    set_seed(); m = Hybrid(mode).to(DEVICE)
    q_params  = list(m.qbank.q_weights)
    cl_params = [p for n_, p in m.named_parameters() if "qbank.q_weights" not in n_]
    ocl = torch.optim.AdamW(cl_params, lr=LR); oq = torch.optim.AdamW(q_params, lr=LR)
    anchor = m.qbank.flat_weights().clone()
    stale_grads = None
    nref = nstep = lastref = 0; ntau = 0; t0 = time.time()
    for ep in range(epochs):
        m.train()
        for x, y in train_loader:
            x, y = x.to(DEVICE), y.to(DEVICE); lat = m.latent(x)
            if lazy:
                drift   = (m.qbank.flat_weights() - anchor).norm().item()
                tau_hit = drift > tau
                forced  = (nstep == 0) or ((nstep - lastref) >= REFRESH_EVERY)
                do_ref  = (nref < MAX_REFRESH) and (forced or tau_hit)
            else:
                do_ref = True                                     # naive: refresh every step
            if do_ref:
                if lazy and tau_hit and not forced: ntau += 1
                for p in q_params: p.requires_grad_(True)
                loss = F.cross_entropy(m.classify(lat, m.quantum(lat)), y)
                ocl.zero_grad(); oq.zero_grad(); loss.backward()
                stale_grads = [p.grad.detach().clone() for p in q_params]  # cache fresh grads
                ocl.step(); oq.step()
                anchor = m.qbank.flat_weights().clone(); nref += 1; lastref = nstep
            else:
                # classical path trains normally; circuit output is a constant feature
                qf = m.quantum(lat, detach_circuit=True)
                loss = F.cross_entropy(m.classify(lat, qf), y)
                ocl.zero_grad(); loss.backward(); ocl.step()
                if stale_grads is not None:                        # stale-gradient q step
                    for p, g_ in zip(q_params, stale_grads): p.grad = g_.clone()
                    oq.step()
            nstep += 1
    acc, f1 = evaluate(m, hybrid=True); cr = contribution_ratio(m)
    if verbose:
        extra = f" tau-hits={ntau}" if lazy else ""
        print(f"    {mode:<16} lazy={str(lazy):<5} tau={tau:<5} -> acc={acc:.4f} f1={f1:.4f} "
              f"Q-bwd={nref}/{nstep}{extra} contrib={cr:.3f} ({time.time()-t0:.0f}s)")
    return dict(acc=acc, f1=f1, t=time.time()-t0, qb=nref, steps=nstep, contrib=cr,
                tau_hits=(ntau if lazy else None), model=m)

# -- 8. MAIN RUNS (QuCNet Table-4 style: w/o vs w/ quantum) --------------------------------
print("\n" + "="*88 + "\n  MAIN — without-TQC vs with-TQC (QuCNet Table-4 style) + lazy vs naive\n" + "="*88)
res_cls   = train_classical()
res_naive = train_hybrid("hcws", lazy=False)          # quantum backward every step
res_lazy  = train_hybrid("hcws", tau=TAU_MAIN)        # ours (stale-gradient lazy)

# -- 9. ABLATION 1 — weight sharing (QuCNet Table-5; NAIVE so circuits fully train) ----------
print("\n" + "="*88 + "\n  ABLATION 1 — weight-sharing strategies (QuCNet Table-5, naive/full)\n" + "="*88)
share_res = {m_: train_hybrid(m_, lazy=False, epochs=ABLATION_EPOCHS) for m_ in SHARE_MODES}

# -- 10. ABLATION 2 — TAU sweep (our headline: acc vs Q-backward passes) --------------------
print("\n" + "="*88 + "\n  ABLATION 2 — TAU sweep (headline: accuracy vs quantum-backward passes)\n" + "="*88)
tau_res = {t_: train_hybrid("hcws", tau=t_, epochs=ABLATION_EPOCHS) for t_ in TAU_SWEEP}

# -- 11. FINAL TABLES -----------------------------------------------------------------------
line = "  " + "-"*84
print("\n" + "="*88 + "\n  RESULTS SUMMARY\n" + "="*88)
print("  [A] With/without quantum + lazy vs naive (main, %d epochs)" % MAIN_EPOCHS)
print(f"  {'Model':<26}{'Acc':>8}{'F1':>8}{'Q-bwd':>8}{'Contrib':>9}{'Time(s)':>9}")
print(line)
print(f"  {'Classical (w/o TQC)':<26}{res_cls['acc']:>8.4f}{res_cls['f1']:>8.4f}{0:>8}{'—':>9}{res_cls['t']:>9.0f}")
print(f"  {'Hybrid naive (every-step)':<26}{res_naive['acc']:>8.4f}{res_naive['f1']:>8.4f}"
      f"{res_naive['qb']:>8}{res_naive['contrib']:>9.3f}{res_naive['t']:>9.0f}")
print(f"  {'Hybrid LAZY (ours)':<26}{res_lazy['acc']:>8.4f}{res_lazy['f1']:>8.4f}"
      f"{res_lazy['qb']:>8}{res_lazy['contrib']:>9.3f}{res_lazy['t']:>9.0f}")
sav = 100*(1 - res_lazy['qb']/max(res_naive['qb'],1))
print(line)
print(f"  Lazy vs naive: {sav:.1f}% fewer quantum backward passes | "
      f"d_acc(lazy-naive)={res_lazy['acc']-res_naive['acc']:+.4f} | "
      f"tau-triggered refreshes={res_lazy['tau_hits']}")

print("\n  [B] Weight-sharing ablation (QuCNet Table-5 style, naive, %d epochs)" % ABLATION_EPOCHS)
print(f"  {'Scheme':<20}{'#sets':>7}{'Acc':>8}{'F1':>8}{'Q-bwd':>8}")
print(line)
for m_, r_ in share_res.items():
    nsets = {"all_shared":1, "all_independent":K_CIRCUITS}.get(m_, R_UNIQUE)
    print(f"  {m_:<20}{nsets:>7}{r_['acc']:>8.4f}{r_['f1']:>8.4f}{r_['qb']:>8}")

print("\n  [C] TAU sweep (headline curve, %d epochs)" % ABLATION_EPOCHS)
print(f"  {'tau':>6}{'Acc':>9}{'F1':>9}{'Q-bwd':>8}{'tau-hits':>9}{'% saved vs naive':>18}")
print(line)
for t_, r_ in tau_res.items():
    print(f"  {t_:>6.2f}{r_['acc']:>9.4f}{r_['f1']:>9.4f}{r_['qb']:>8}{r_['tau_hits']:>9}"
          f"{100*(1-r_['qb']/max(r_['steps'],1)):>17.1f}%")
print("="*88)
print("\n[read] tau-hits>0 => the trigger is live; Q-bwd must now vary with tau, and small-tau")
print("       runs should approach naive accuracy. If tau-hits==0, stale grads too weak: raise LR_q.")
print("[next] raise ABLATION_EPOCHS + add Fashion-MNIST/CIFAR-10 for the paper.")
```

```python
# ======================================================================================
# LazyQ-HRU v5.1  |  FULL MNIST, one cell
# --------------------------------------------------------------------------------------
# Arms: Classical CNN | Naive hybrid (quantum backward every step) | LAZY (ours)
# Simplified circuit: 4 qubits, K=4 parallel circuits, R=2 cyclic sharing (HCWS-lite),
#                     RU=2 re-uploading layers, <Z> expval measurement (QDIM=4).
#
# v5.1: sanity-3 synthetic grad 0.01 -> 0.1 (0.01 gives ~0.06 total drift over 30 steps,
#       mathematically below tau=0.15 -> the trigger could never fire in the test).
#
# v5 changes vs v4 (each grounded in the v4 log):
#   (a) MULTI-SEED: every config runs SEEDS=[7,17,27]; tables report mean +/- std.
#       v4's tau sweep direction (acc UP with tau) was single-seed noise; this kills it.
#   (b) SMALLER STALENESS CAP + DECAY: REFRESH_EVERY 100 -> 25; stale grads applied as
#       g * GAMMA^age (GAMMA=0.9, age = steps since refresh) so stale directions fade.
#       REQUIRED CHANGE: quantum optimizer is plain SGD (no momentum) in ALL arms —
#       Adam's adaptive normalization cancels the decay factor and its momentum state
#       is exactly what compounded stale directions in v4. Classical side stays AdamW.
#   (c) EMA STALE GRADS: fresh grads are folded into an EMA across refreshes
#       (BETA=0.5) instead of reusing one snapshot.
#   + STALE-MODE ABLATION {frozen, snapshot, ema_decay} at tau=0.15:
#       frozen   = v3 pathology control (no stale steps -> tau inert, forced-only)
#       snapshot = v4 behavior (last fresh grad, undecayed)
#       ema_decay= v5 (ours)
# All method-layer framing/results need supervisor sign-off before circulation.
# ======================================================================================

import os, math, time, random, warnings
warnings.filterwarnings("ignore")
import numpy as np
import torch, torch.nn as nn, torch.nn.functional as F
from torch.utils.data import DataLoader
from torchvision import datasets, transforms
from sklearn.metrics import accuracy_score, f1_score
import pennylane as qml

# -- 1. REPRO ---------------------------------------------------------------------------
def set_seed(s):
    random.seed(s); np.random.seed(s); torch.manual_seed(s); torch.cuda.manual_seed_all(s)
DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"[env] device={DEVICE} | torch={torch.__version__} | pennylane={qml.__version__}")

# -- 2. CONFIG ---------------------------------------------------------------------------
SEEDS                 = [7, 17, 27]          # (a) multi-seed
BATCH_SIZE            = 128
LR_CL, LR_Q           = 1e-3, 5e-2           # AdamW classical | plain SGD quantum
LATENT_DIM, N_CLASSES = 16, 10

N_QUBITS, K_CIRCUITS, R_UNIQUE, RU_LAYERS = 4, 4, 2, 2   # simplified circuit
QDIM = N_QUBITS                                          # expval <Z> per qubit

TAU_MAIN      = 0.15
REFRESH_EVERY = 25          # (b) hard staleness cap (was 100)
MAX_REFRESH   = 10**9       # no artificial cap; tau + cap govern
GAMMA         = 0.9         # (b) stale-grad decay per step of age
BETA          = 0.5         # (c) EMA coefficient for fresh grads across refreshes

EPOCHS        = 3           # full MNIST converges fast; 469 steps/epoch @ bs=128
TAU_SWEEP     = [0.05, 0.15, 0.40]
STALE_MODES   = ["frozen", "snapshot", "ema_decay"]

print(f"[cfg] qubits={N_QUBITS} K={K_CIRCUITS} R={R_UNIQUE} RU={RU_LAYERS} qdim={QDIM} "
      f"| tau={TAU_MAIN} cap={REFRESH_EVERY} gamma={GAMMA} beta={BETA} "
      f"| seeds={SEEDS} epochs={EPOCHS}")
print("[cfg] NOTE: quantum optimizer = plain SGD (all arms). Adam's normalization cancels")
print("      gamma-decay and its momentum compounds stale directions (the v4 failure).")

# -- 3. CIRCUIT (re-uploading, RY encoding, ring CNOT, <Z> readout) -----------------------
# default.qubit + backprop: vectorized over the batch, exact gradients.
_dev = qml.device("default.qubit", wires=N_QUBITS)

@qml.qnode(_dev, interface="torch", diff_method="backprop")
def circuit(inputs, weights):
    for l in range(RU_LAYERS):
        qml.AngleEmbedding(inputs[:, l*N_QUBITS:(l+1)*N_QUBITS],
                           wires=range(N_QUBITS), rotation="Y")     # fresh block / layer
        for q in range(N_QUBITS): qml.RY(weights[l, q], wires=q)
        for q in range(N_QUBITS-1): qml.CNOT(wires=[q, q+1])
        qml.CNOT(wires=[N_QUBITS-1, 0])
    return [qml.expval(qml.PauliZ(q)) for q in range(N_QUBITS)]

def run_circuit(inputs, weights):
    # simulator state lives on CPU: inputs+weights must be CPU (differentiable .cpu());
    # caller ships outputs back to the model device.
    out = circuit(inputs.to("cpu"), weights.to("cpu"))
    return torch.stack(out, dim=1).float()                           # [B, N_QUBITS]

# -- 4. MODEL ------------------------------------------------------------------------------
class TinyCNN(nn.Module):
    def __init__(self, d=LATENT_DIM):
        super().__init__()
        self.body = nn.Sequential(
            nn.Conv2d(1, 8, 3, padding=1),  nn.ReLU(), nn.MaxPool2d(2),
            nn.Conv2d(8, 16, 3, padding=1), nn.ReLU(), nn.MaxPool2d(2),
            nn.AdaptiveAvgPool2d(1))
        self.proj = nn.Linear(16, d)
    def forward(self, x): return torch.tanh(self.proj(self.body(x).flatten(1)))

class Head(nn.Module):
    def __init__(self, ind):
        super().__init__()
        self.net = nn.Sequential(nn.Linear(ind, 64), nn.LayerNorm(64), nn.GELU(),
                                 nn.Dropout(0.1), nn.Linear(64, N_CLASSES))
    def forward(self, x): return self.net(x)

def share_id(k): return k % R_UNIQUE                     # HCWS stride

class QBank(nn.Module):
    def __init__(self, d=LATENT_DIM):
        super().__init__()
        fpc = RU_LAYERS * N_QUBITS
        self.adapters  = nn.ModuleList([nn.Linear(d, fpc) for _ in range(K_CIRCUITS)])
        self.q_weights = nn.ParameterList([nn.Parameter(0.1*torch.randn(RU_LAYERS, N_QUBITS))
                                           for _ in range(R_UNIQUE)])
    def forward(self, latent, detach_circuit=False):
        outs = []
        for k in range(K_CIRCUITS):
            feats = torch.tanh(self.adapters[k](latent)) * math.pi
            if detach_circuit:
                with torch.no_grad():
                    zk = run_circuit(feats.detach(), self.q_weights[share_id(k)])
            else:
                zk = run_circuit(feats, self.q_weights[share_id(k)])
            outs.append(zk.to(latent.device))
        return torch.cat(outs, dim=1)                    # [B, K*QDIM]
    def flat_weights(self):
        return torch.cat([w.detach().reshape(-1) for w in self.q_weights])

class Hybrid(nn.Module):
    def __init__(self):
        super().__init__()
        self.cnn = TinyCNN(); self.qbank = QBank()
        self.head = Head(LATENT_DIM + K_CIRCUITS * QDIM)
    def latent(self, x): return self.cnn(x)
    def quantum(self, l, detach_circuit=False): return self.qbank(l, detach_circuit)
    def classify(self, l, q): return self.head(torch.cat([l, q], dim=1))

class Classical(nn.Module):
    def __init__(self):
        super().__init__(); self.cnn = TinyCNN(); self.head = Head(LATENT_DIM)

# -- 5. DATA (FULL MNIST) --------------------------------------------------------------------
print("\n[data] loading FULL MNIST ...")
tf = transforms.Compose([transforms.ToTensor(), transforms.Normalize((0.1307,), (0.3081,))])
_tr = datasets.MNIST("./data", train=True,  download=True, transform=tf)
_te = datasets.MNIST("./data", train=False, download=True, transform=tf)
train_loader = DataLoader(_tr, batch_size=BATCH_SIZE, shuffle=True,  num_workers=2, pin_memory=True)
test_loader  = DataLoader(_te, batch_size=256,        shuffle=False, num_workers=2, pin_memory=True)
STEPS_PER_EPOCH = len(train_loader)
print(f"[data] train={len(_tr)} test={len(_te)} batch={BATCH_SIZE} steps/epoch={STEPS_PER_EPOCH}")

# -- 6. SANITY CHECKS ------------------------------------------------------------------------
print("\n[sanity-1] circuit shape + gradient flow ...")
set_seed(SEEDS[0])
_b = QBank().to(DEVICE)
_l = torch.randn(BATCH_SIZE, LATENT_DIM, device=DEVICE, requires_grad=True)
_t = time.time(); _q = _b(_l); _dt = time.time() - _t
assert _q.shape == (BATCH_SIZE, K_CIRCUITS * QDIM), _q.shape
_q.sum().backward()
assert all(w.grad is not None and w.grad.abs().sum() > 0 for w in _b.q_weights), "no q-grad!"
assert _l.grad is not None and _l.grad.abs().sum() > 0, "no grad to latent!"
print(f"          out={tuple(_q.shape)} fwd={_dt*1000:.0f}ms | q-grads OK, latent-grad OK | "
      f"share map={[share_id(k) for k in range(K_CIRCUITS)]}")

print("[sanity-1b] detached path: circuit output constant, classical grads intact ...")
_l2 = torch.randn(BATCH_SIZE, LATENT_DIM, device=DEVICE, requires_grad=True)
for w in _b.q_weights: w.grad = None
_q2 = _b(_l2, detach_circuit=True)
assert not _q2.requires_grad, "detached circuit output must carry no graph"
del _b, _l, _q, _l2, _q2

print("\n[sanity-2] Lipschitz premise (tau-dial reliability) ...")
set_seed(SEEDS[0])
_p = Hybrid().to(DEVICE).eval()
_xb = next(iter(test_loader))[0][:BATCH_SIZE].to(DEVICE)
with torch.no_grad():
    _l0 = _p.latent(_xb); _q0 = _p.quantum(_l0)
    _w0 = [w.clone() for w in _p.qbank.q_weights]
    print(f"          {'eps':>6} | {'||dW||':>8} | {'||dFeat||':>9} | {'ratio~Lq':>8}")
    for eps in [0.01, 0.1, 0.5]:
        for w, wo in zip(_p.qbank.q_weights, _w0): w.copy_(wo + eps*torch.randn_like(wo))
        dW = math.sqrt(sum(((w-wo)**2).sum().item() for w,wo in zip(_p.qbank.q_weights,_w0)))
        dF = (_p.quantum(_l0) - _q0).norm().item()
        print(f"          {eps:>6.2f} | {dW:>8.3f} | {dF:>9.3f} | {dF/max(dW,1e-9):>8.3f}")
    for w, wo in zip(_p.qbank.q_weights, _w0): w.copy_(wo)
del _p

print("\n[sanity-3] stale-grad machinery closed-loop (synthetic, 30 steps) ...")
# Calibration: 16 q-params, SGD step = LR_Q * g * decay. With g=0.1 the drift norm grows
# ~sqrt(16)*LR_Q*0.1*avg_decay ~ 0.016/step -> first tau-hit ~step 10, 2-3 hits in 30.
# (g=0.01 tops out at ~0.06 total drift: below tau by construction — v5's test bug.)
set_seed(SEEDS[0])
_m = Hybrid().to(DEVICE)
_qp = list(_m.qbank.q_weights)
_oq = torch.optim.SGD(_qp, lr=LR_Q)
_anchor = _m.qbank.flat_weights().clone()
_g = [0.1*torch.ones_like(p) for p in _qp]      # fixed synthetic stale grad (see calib.)
_hits, _max_drift, _last = 0, 0.0, 0
for _s in range(30):
    drift = (_m.qbank.flat_weights() - _anchor).norm().item()
    _max_drift = max(_max_drift, drift)
    if drift > TAU_MAIN:
        _hits += 1; _anchor = _m.qbank.flat_weights().clone(); _last = _s
    for p, g_ in zip(_qp, _g): p.grad = g_ * (GAMMA ** ((_s - _last) if _hits else _s % 5))
    _oq.step()
assert _hits > 0, (f"tau trigger never fired: max drift {_max_drift:.4f} vs tau {TAU_MAIN} "
                   f"— synthetic grad / LR_Q / tau are inconsistent")
_post = (_m.qbank.flat_weights() - _anchor).norm().item()
print(f"          tau-hits={_hits}/30 | max drift={_max_drift:.3f} vs tau={TAU_MAIN} | "
      f"drift since last refresh={_post:.3f} | decay @age5={GAMMA**5:.3f}")
del _m

print("\n[sanity-4] determinism: same seed -> identical first-batch loss ...")
_losses = []
for _rep in range(2):
    set_seed(SEEDS[0])
    _mm = Hybrid().to(DEVICE)
    _xb2, _yb2 = next(iter(DataLoader(_tr, batch_size=BATCH_SIZE, shuffle=True,
                                      generator=torch.Generator().manual_seed(SEEDS[0]))))
    _xb2, _yb2 = _xb2.to(DEVICE), _yb2.to(DEVICE)
    _losses.append(F.cross_entropy(_mm.classify(_mm.latent(_xb2),
                                                _mm.quantum(_mm.latent(_xb2))), _yb2).item())
    del _mm
assert abs(_losses[0] - _losses[1]) < 1e-6, f"non-deterministic init: {_losses}"
print(f"          loss rep1={_losses[0]:.6f} rep2={_losses[1]:.6f} (identical)")

# -- 7. TRAIN / EVAL --------------------------------------------------------------------------
def evaluate(model, hybrid):
    model.eval(); ys, ps = [], []
    with torch.no_grad():
        for x, y in test_loader:
            x = x.to(DEVICE)
            logits = (model.classify(model.latent(x), model.quantum(model.latent(x)))
                      if hybrid else model.head(model.cnn(x)))
            ps.append(logits.argmax(1).cpu().numpy()); ys.append(y.numpy())
    ys, ps = np.concatenate(ys), np.concatenate(ps)
    return accuracy_score(ys, ps), f1_score(ys, ps, average="macro")

def train_classical(seed, verbose=True):
    set_seed(seed); m = Classical().to(DEVICE)
    opt = torch.optim.AdamW(m.parameters(), lr=LR_CL); t0 = time.time()
    for ep in range(EPOCHS):
        m.train()
        for x, y in train_loader:
            x, y = x.to(DEVICE), y.to(DEVICE)
            loss = F.cross_entropy(m.head(m.cnn(x)), y)
            opt.zero_grad(); loss.backward(); opt.step()
    acc, f1 = evaluate(m, hybrid=False); t = time.time() - t0
    if verbose: print(f"    classical        seed={seed:<3} acc={acc:.4f} f1={f1:.4f} ({t:.0f}s)")
    del m; torch.cuda.empty_cache()
    return dict(acc=acc, f1=f1, t=t, qb=0, steps=EPOCHS*STEPS_PER_EPOCH, tau_hits=0)

def train_hybrid(seed, lazy=True, tau=TAU_MAIN, stale_mode="ema_decay", verbose=True):
    """lazy=False -> naive: quantum backward every step.
    lazy=True stale_mode:
      'frozen'    : no stale q-steps (v3 pathology control: tau inert, forced-only refreshes)
      'snapshot'  : stale grads applied undecayed (v4 behavior)
      'ema_decay' : EMA of fresh grads (BETA) applied with GAMMA^age decay (v5, ours)"""
    set_seed(seed); m = Hybrid().to(DEVICE)
    q_params  = list(m.qbank.q_weights)
    cl_params = [p for n_, p in m.named_parameters() if "qbank.q_weights" not in n_]
    ocl = torch.optim.AdamW(cl_params, lr=LR_CL)
    oq  = torch.optim.SGD(q_params, lr=LR_Q)             # plain SGD: see cfg note
    anchor = m.qbank.flat_weights().clone()
    ema_grads = None
    nref = nstep = lastref = ntau = 0; t0 = time.time()
    for ep in range(EPOCHS):
        m.train()
        for x, y in train_loader:
            x, y = x.to(DEVICE), y.to(DEVICE); lat = m.latent(x)
            if lazy:
                drift   = (m.qbank.flat_weights() - anchor).norm().item()
                tau_hit = drift > tau
                forced  = (nstep == 0) or ((nstep - lastref) >= REFRESH_EVERY)
                do_ref  = (nref < MAX_REFRESH) and (forced or tau_hit)
            else:
                do_ref = True
            if do_ref:
                if lazy and tau_hit and not forced: ntau += 1
                for p in q_params: p.requires_grad_(True)
                loss = F.cross_entropy(m.classify(lat, m.quantum(lat)), y)
                ocl.zero_grad(); oq.zero_grad(); loss.backward()
                fresh = [p.grad.detach().clone() for p in q_params]
                ema_grads = (fresh if ema_grads is None else
                             [BETA*e + (1-BETA)*g for e, g in zip(ema_grads, fresh)])
                ocl.step(); oq.step()
                anchor = m.qbank.flat_weights().clone(); nref += 1; lastref = nstep
            else:
                qf = m.quantum(lat, detach_circuit=True)          # constant feature
                loss = F.cross_entropy(m.classify(lat, qf), y)
                ocl.zero_grad(); loss.backward(); ocl.step()
                if stale_mode != "frozen" and ema_grads is not None:
                    age   = nstep - lastref
                    scale = 1.0 if stale_mode == "snapshot" else GAMMA ** age
                    for p, g_ in zip(q_params, ema_grads): p.grad = g_ * scale
                    oq.step()
            nstep += 1
    acc, f1 = evaluate(m, hybrid=True); t = time.time() - t0
    if verbose:
        tag = "naive" if not lazy else stale_mode
        print(f"    {tag:<16} seed={seed:<3} tau={tau:<5} acc={acc:.4f} f1={f1:.4f} "
              f"Q-bwd={nref}/{nstep} tau-hits={ntau} ({t:.0f}s)")
    del m; torch.cuda.empty_cache()
    return dict(acc=acc, f1=f1, t=t, qb=nref, steps=nstep, tau_hits=ntau)

def agg(runs):
    """mean +/- std over seeds for the fields we report."""
    def ms(key):
        v = np.array([r[key] for r in runs], dtype=float)
        return v.mean(), v.std()
    return {k: ms(k) for k in ["acc", "f1", "qb", "tau_hits", "t"]}

def fmt(m, s, p=4): return f"{m:.{p}f}±{s:.{p}f}"

# -- 8. RUNS ----------------------------------------------------------------------------------
TOTAL_STEPS = EPOCHS * STEPS_PER_EPOCH
print(f"\n[plan] {len(SEEDS)} seeds x (1 classical + 1 naive + {len(STALE_MODES)} stale-modes "
      f"+ {len(TAU_SWEEP)-1} extra taus) hybrid runs; naive is the slow one "
      f"({TOTAL_STEPS} quantum backwards/run). Expect ~45-60 min on a T4.")

print("\n" + "="*88 + "\n  [A] MAIN — classical vs naive vs lazy(ema_decay, tau=0.15), full MNIST\n" + "="*88)
runs_cls   = [train_classical(s) for s in SEEDS]
runs_naive = [train_hybrid(s, lazy=False) for s in SEEDS]

print("\n" + "="*88 + "\n  [B] STALE-MODE ABLATION — frozen (v3) | snapshot (v4) | ema_decay (v5), tau=0.15\n" + "="*88)
runs_mode = {md: [train_hybrid(s, lazy=True, tau=TAU_MAIN, stale_mode=md) for s in SEEDS]
             for md in STALE_MODES}
runs_lazy = runs_mode["ema_decay"]                        # doubles as main lazy arm

print("\n" + "="*88 + "\n  [C] TAU SWEEP — ema_decay, multi-seed (tau=0.15 reused from [B])\n" + "="*88)
runs_tau = {TAU_MAIN: runs_lazy}
for t_ in TAU_SWEEP:
    if t_ == TAU_MAIN: continue
    runs_tau[t_] = [train_hybrid(s, lazy=True, tau=t_, stale_mode="ema_decay") for s in SEEDS]
runs_tau = dict(sorted(runs_tau.items()))

# -- 9. FINAL TABLES ---------------------------------------------------------------------------
line = "  " + "-"*86
print("\n" + "="*88 + f"\n  RESULTS (mean±std over seeds {SEEDS}, {EPOCHS} epochs, full MNIST)\n" + "="*88)

a_cls, a_nv, a_lz = agg(runs_cls), agg(runs_naive), agg(runs_lazy)
print("  [A] Main comparison")
print(f"  {'Model':<28}{'Acc':>16}{'F1':>16}{'Q-bwd':>14}{'Time(s)':>12}")
print(line)
print(f"  {'Classical (no quantum)':<28}{fmt(*a_cls['acc']):>16}{fmt(*a_cls['f1']):>16}"
      f"{'0':>14}{fmt(*a_cls['t'],0):>12}")
print(f"  {'Hybrid naive (every step)':<28}{fmt(*a_nv['acc']):>16}{fmt(*a_nv['f1']):>16}"
      f"{fmt(*a_nv['qb'],0):>14}{fmt(*a_nv['t'],0):>12}")
print(f"  {'Hybrid LAZY (ours)':<28}{fmt(*a_lz['acc']):>16}{fmt(*a_lz['f1']):>16}"
      f"{fmt(*a_lz['qb'],0):>14}{fmt(*a_lz['t'],0):>12}")
print(line)
sav = 100*(1 - a_lz['qb'][0]/max(a_nv['qb'][0],1))
print(f"  Lazy vs naive: {sav:.1f}% fewer quantum backwards | "
      f"d_acc = {a_lz['acc'][0]-a_nv['acc'][0]:+.4f} "
      f"(seed-std: lazy {a_lz['acc'][1]:.4f}, naive {a_nv['acc'][1]:.4f})")

print("\n  [B] Stale-mode ablation (tau=0.15)  -> does (b)+(c) close the v4 gap?")
print(f"  {'Mode':<14}{'Acc':>16}{'F1':>16}{'Q-bwd':>14}{'tau-hits':>14}")
print(line)
for md in STALE_MODES:
    a = agg(runs_mode[md])
    print(f"  {md:<14}{fmt(*a['acc']):>16}{fmt(*a['f1']):>16}"
          f"{fmt(*a['qb'],0):>14}{fmt(*a['tau_hits'],0):>14}")

print("\n  [C] TAU sweep (ema_decay)  -> headline: acc vs quantum-backwards, noise-controlled")
print(f"  {'tau':>6}{'Acc':>16}{'F1':>16}{'Q-bwd':>14}{'tau-hits':>12}{'% saved':>10}")
print(line)
for t_, rr in runs_tau.items():
    a = agg(rr)
    print(f"  {t_:>6.2f}{fmt(*a['acc']):>16}{fmt(*a['f1']):>16}{fmt(*a['qb'],0):>14}"
          f"{fmt(*a['tau_hits'],0):>12}{100*(1-a['qb'][0]/TOTAL_STEPS):>9.1f}%")
print("="*88)
print("\n[read] (1) [B]: ema_decay must beat frozen and snapshot, else (b)+(c) didn't help.")
print("       (2) [C]: with seed-std bars, acc should now be flat-or-decreasing in tau;")
print("           if error bars overlap fully across tau, staleness cost is < noise here.")
print("       (3) [A]: the efficiency claim needs d_acc within ~1 std of zero at >90% savings.")
print("[next] if [A] gap persists: raise refresh budget (cap=10) or per-sample feature cache.")
```
